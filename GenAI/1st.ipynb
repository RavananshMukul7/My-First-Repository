```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("california-housing - california-housing.csv")
df = df.drop_duplicates()
df = df.dropna()
af = df.head(10)
# x = af["total_bedrooms"]
# y = af["median_house_value"]
# y = y.astype(float)
# x.head()
# y.dtype
```

```python
x = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])
w = 0
b = 0
n = len(x)
epochs = 1000
a = 0.01

for epoch in range(epochs + 1):
  p_y = w * x + b
  
  loss = np.sum((y - p_y)**2)/n
  dw = -(2/n)*np.sum(x*(y - p_y))
  db = -(2/n)*np.sum(y - p_y)

  w = w - a*dw
  b = b - a*db

  if epoch % 100 == 0:
    print(f"Epoch {epoch}: Loss={loss}, w={w}, b={b}")

print(f"Final Weight: {w:.4f}")
print(f"Final Bias: {b:.4f}")
```
